# Iris Dataset Analysis with Pandas and Matplotlib

This notebook demonstrates how to analyze and visualize the Iris dataset using Python libraries such as Pandas, Matplotlib, and Seaborn.

```python
import pandas as pd # Import Pandas for data manipulation 

# Load the dataset
df = pd.read_csv('/kaggle/input/iris/Iris.csv') # Replace with the correct file path
df.head()   # Display the first 5 rows of the dataset
```

```python
# Check the data types of each column and look for missing values
print(df.dtypes)
print("\nMissing values:\n", df.isnull().sum())
```

```python
# Get summary statistics for numerical columns
df.describe()
```

```python
import matplotlib.pyplot as plt # Import Matplotlib for plotting
import seaborn as sns # Import Seaborn for advanced visualizations
```

```python
# Line chart: Simulate a time series using index
df['Index'] = df.index # Add an index column to simulate time series
plt.plot(df['Index'], df['SepalLengthCm'])  # Plot Sepal Length against the index
plt.title("Sepal Length Over Time (Index)") # Add a title
plt.xlabel("Index") # Label the x-axis
plt.ylabel("Sepal Length (cm)") # Label the y-axis
plt.show() # Display the plot
```

```python
# Group data by species and calculate the mean for each group
grouped = df.groupby('Species').mean()
```

```python
# Bar chart: Average petal length per species
grouped['PetalLengthCm'].plot(kind='bar', color='skyblue')  # Plot a bar chart
plt.title("Average Petal Length per Species") # Add a title
plt.ylabel("Petal Length (cm)") # Label the y-axis
plt.xlabel("Species") # Label the x-axis
plt.show() # Display the plot
```

```python
# Histogram: Distribution of sepal width
plt.hist(df['SepalWidthCm'], bins=15, color='lightgreen', edgecolor='black') # Create a histogram
plt.title("Distribution of Sepal Width") # Add a title
plt.xlabel("Sepal Width (cm)") # Label the x-axis
plt.ylabel("Frequency") # Label the y-axis
plt.show() # Display the plot
```

```python
# Scatter plot: Sepal Length vs Petal Length
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], alpha=0.7, c='coral')  # Create a scatter plot
plt.title("Sepal Length vs Petal Length") # Add a title
plt.xlabel("Sepal Length (cm)") # Label the x-axis
plt.ylabel("Petal Length (cm)") # Label the y-axis
plt.show() # Display the plot
```

### 📌 Observations:

- The Setosa species has the smallest petal length on average.
- Sepal width appears to follow a normal distributed.
- There is a clear positive correlation between Sepal Length and Petal Length.
